```python
import pickle
from tqdm import tqdm
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
from data.movie import Frames
from data.utils import l1, l2, ssim, wasserstein, correlation, cosine
```

```python
frames = Frames().raw
frames.shape
```

```python
# Organizing our metrics and stores
measures = {}
metrics = [# On pure pixel traces
           l1, l2,
           correlation, cosine,
           
           # On population of Pixels
           ssim, wasserstein]

# Making measurements
for method in metrics:
    key = method.__name__
    measures[key] = []
    for i in tqdm(range(len(frames) - 1), desc=key):
        frame_1 = frames[..., 0][i]
        frame_2 = frames[..., 0][i+1]
        measures[key].append(method(frame_1, frame_2))
```

```python
df = pd.DataFrame(measures)
df
```
