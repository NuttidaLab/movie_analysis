```python
min_max_norm = lambda x: (x - x.min()) / (x.max() - x.min())
l2 = min_max_norm(frames.measures.l2.rescale(1))
ss = min_max_norm(susp.rescale(1).mean(axis=-1))
tm = min_max_norm(audio.instruments.timbre.rescale(1))

plt.figure( figsize=(15,3) )
plt.plot(l2, "k-", label="Frame L2")
plt.plot(ss, "r-", label="Suspense")
plt.plot(tm, "b-", label="Timbre")
plt.legend()
plt.show()
```
